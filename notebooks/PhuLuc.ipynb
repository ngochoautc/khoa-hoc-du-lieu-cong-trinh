## PL.4. PHÂN PHỐI MẪU VÀ KHOẢNG TIN CẬY

```python
import numpy as np
from scipy import stats

# cuong do nen 10 mau (MPa), so lieu gia dinh
fc = np.array([38.2, 41.5, 36.9, 40.1, 39.4,
               42.3, 37.8, 40.8, 38.9, 41.0])
n = len(fc)
xtb, s = fc.mean(), fc.std(ddof=1)
print(f"n = {n}, x_tb = {xtb:.2f} MPa, s = {s:.2f} MPa")
# khoang tin cay 95% cho trung binh (sigma chua biet)
lo, hi = stats.t.interval(0.95, df=n - 1, loc=xtb,
                          scale=s / np.sqrt(n))
print(f"t_0.025 = {stats.t.ppf(0.975, n - 1):.3f}")
print(f"KTC 95% cua mu   : [{lo:.2f}; {hi:.2f}] MPa")
# khoang tin cay 95% cho do lech chuan
c_tren = stats.chi2.ppf(0.975, n - 1)
c_duoi = stats.chi2.ppf(0.025, n - 1)
s_lo = np.sqrt((n - 1) * s**2 / c_tren)
s_hi = np.sqrt((n - 1) * s**2 / c_duoi)
print(f"chi2: {c_duoi:.3f} va {c_tren:.3f}")
print(f"KTC 95% cua sigma: [{s_lo:.2f}; {s_hi:.2f}] MPa")
```

## PL.5. KIỂM ĐỊNH GIẢ THUYẾT

```python
# cuong do 28 ngay cua hai tram tron (MPa), so lieu mo phong
rng = np.random.default_rng(12)
fA = rng.normal(38.0, 2.5, 15).round(1)
fB = rng.normal(36.0, 2.5, 15).round(1)
for ten, x in [("A", fA), ("B", fB)]:
    W, p = stats.shapiro(x)
    print(f"Tram {ten}: TB = {x.mean():.2f} MPa, "
          f"s = {x.std(ddof=1):.2f} MPa, "
          f"Shapiro W = {W:.3f}, p = {p:.3f}")
# kiem dinh t hai mau dang Welch
kq = stats.ttest_ind(fA, fB, equal_var=False)
print(f"Welch: t = {kq.statistic:.3f}, "
      f"bac tu do = {kq.df:.1f}, p = {kq.pvalue:.4f}")
```
